```python
df = df.fillna(df.median(numeric_only=True))
```

```python
df = df.drop("CUST_ID",axis=1)
```

```python
df
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
scaled_data = scaler.fit_transform(df)
```

```python
df.head()
```

```python
import matplotlib.pyplot as plt
```

```python
from sklearn.cluster import KMeans
wcss = []
for k in range(1,11):
    kmeans = KMeans(n_clusters=k,random_state=42)
    kmeans.fit(scaled_data)
    wcss.append(kmeans.inertia_)
plt.plot(range(1,11),wcss)
plt.xlabel("Number of clusters")
plt.ylabel("wcss")
plt.title("Elbow method")
plt.show()
```

```python
kmeans = KMeans(n_clusters=4,random_state=42)
clusters = kmeans.fit_predict(scaled_data)
df["clusters"] = clusters
df.head()
```

```python
from sklearn.metrics import silhouette_score
score = silhouette_score(scaled_data,clusters)
print("Silhouette Score:",score)
```

```python
from sklearn.decomposition import PCA
pca = PCA(n_components=2)
pca_data = pca.fit_transform(scaled_data)
```

```python
df["PCA1"] = pca_data[:,0]
df["PCA2"] = pca_data[:,1]
```

```python
import seaborn as sns
sns.scatterplot(
    x="PCA1",y="PCA2",
    hue="clusters",
    data=df,
    palette="Set2"
)
```

```python
from sklearn.metrics import silhouette_score
silhouette_score(scaled_data, df["clusters"])
```

```python
for k in range(2,11):
    kmeans = KMeans(n_clusters=k,random_state=42)
    labels=kmeans.fit_predict(scaled_data)
    score = silhouette_score(scaled_data,labels)
    print(f"K={k}:Silhoutte={score:.4f}")
```

```python
from sklearn.cluster import DBSCAN

db = DBSCAN(eps=1.5, min_samples=5)
labels = db.fit_predict(scaled_data)

silhouette_score(scaled_data, labels)
```

```python
from sklearn.decomposition import PCA
import seaborn as sns
import matplotlib.pyplot as plt

pca = PCA(n_components=2)
pca_data = pca.fit_transform(scaled_data)

df["PCA1"] = pca_data[:,0]
df["PCA2"] = pca_data[:,1]

sns.scatterplot(x="PCA1", y="PCA2", hue="clusters", data=df)
plt.title("Customer Segmentation (K=2)")
plt.show()
```
