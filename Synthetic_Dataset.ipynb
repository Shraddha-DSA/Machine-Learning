```python
import pandas as pd

real_df = pd.DataFrame({
    "age": [25, 32, 45, 29, 38],
    "salary": [50000, 75000, 120000, 65000, 90000],
    "department": ["HR", "IT", "Finance", "IT", "Marketing"],
    "experience": [2, 7, 15, 4, 10]
})

print(real_df)
```

```python
from sdv.single_table import CTGANSynthesizer
from sdv.metadata import Metadata
metadata=Metadata.detect_from_dataframe(real_df)
synthesizer=CTGANSynthesizer(metadata)
synthesizer.fit(real_df)
synthetic_df=synthesizer.sample(5000)
```

```python
#SDV learns the relationship between columns(eg age salary etc) aand generate realistic new rows
```

```python
synthetic_df.head()
```

```python
from ctgan import CTGAN

ctgan = CTGAN()

categorical_columns = ["department"]  # replace with your string columns

ctgan.fit(real_df, discrete_columns=categorical_columns)
synthetic=ctgan.sample(1000)
```
